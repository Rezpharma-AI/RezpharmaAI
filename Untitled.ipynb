```python
import xml.etree.ElementTree as ET

def xray_xml(drug_name):
    print(f"Scanning raw XML for {drug_name}... (Please wait ~30 seconds)")
    # Point this to your exact XML path
    context = ET.iterparse("data/raw/full database.xml", events=("end",))
    
    for event, elem in context:
        # Check if this is a <drug> element
        if elem.tag.endswith("}drug") or elem.tag == "drug":
            name = ""
            # Find the name tag
            for child in elem:
                if child.tag.endswith("}name") or child.tag == "name":
                    name = child.text or ""
                    break
            
            # If we found our drug, print its internal structure
            if drug_name.lower() in name.lower():
                print(f"\n✅ Found '{name}'. Here are its actual XML tags:")
                for child in elem:
                    tag = child.tag.split('}')[-1] if '}' in child.tag else child.tag
                    text = (child.text or "")[:100].replace('\n', ' ').strip()
                    print(f"  • <{tag}>: {text}")
                return
                
        # Clear memory to prevent RAM crash
        elem.clear()
        
    print("Could not find drug in XML.")

xray_xml("Amiodarone")
```
